# Funciones

El concepto de función es básico en prácticamente cualquier lenguaje de programación. Se trata de una estructura que nos permite agrupar código. Persigue dos objetivos claros:

1. **No repetir** fragmentos de código en un programa.

1. **Reutilizar** el código en distintos escenarios.

![Descripción de la función](imagenes/funcion.png)

## Primera función

Vamos *invocar* a esta función y obtendremos un mensaje por pantalla, fruto de la ejecución del cuerpo de la función.

````python
def say_hello():
    print("Hello!")

# Esta sería la forma de invocar una función.
say_hello()
````

Las funciones pueden devolver un valor.

````python
def one():
    return 1

print(one())
````

Podemos integrar las funciones de otras estructuras como en condicionales:

````python
if one() == 1:
    print("¡¡¡ Funciona !!!")
else:
    print("Algo va mal :(")
````

Una función puede devolver más de un valor.

````python
def multiple():
    return 0, 1  # Es una tupla.

result = multiple()
print(result)
print(type(result))
````

````python
# Podemos obtener los valores por separado.
a, b = multiple()
print(a)
print(b)
````

## Parámetros y argumentos

Una función sin valores de entrada estaría muy limitada. Sin embargo los parámetros nos permiten variar los datos que usa una función para obtener distintos resultados.

````python
def sqrt(valor):  # valor es un parámetro
    raiz = valor ** (1 / 2)
    return raiz

print(sqrt(4))  # 4 es un argumento
````

````python
def minimo(a, b):
    if a < b:
        return a
    else:
        return b

print(minimo(7, 9))
````

````python
# Una versión más compacta de la misma función.
def minimo(a, b):
    if a < b:
        return a
    return b

print(minimo(7, 9))
````

Tenemos dos tipos de argumentos:

* **Posicionales**: Se copian en sus correspondientes parámetros en orden.
* **Nominales**: Se asignan por nombre a cada parámetro.

**Ejemplo Posicionales**

````python
def build_cpu(vendor, num_cores, freq):
    return {
        "vendor": vendor,
        "num_cores": num_cores,
        "freq": freq,
    }

diccionario_cpu = build_cpu("AMD", 8, 2.7)
print(diccionario_cpu)
````

**Ejemplo Nominales**

````python
print(build_cpu(vendor="AMD", num_cores=8, freq=2.7))
````

````python
# El orden de los argumentos nominales no influye en el resultado.
print(build_cpu(num_cores=8, freq=2.7, vendor="AMD"))
````

Python **permite mezclar argumentos** posicionales y nominales en la llamada a una función:

````python
print(build_cpu("INTEL", num_cores=4, freq=3.1))
````

Los argumentos posicionales deben ir siempre antes.

El siguiente ejemplo es **intencionadamente inválido** porque un argumento posicional no puede ir después de uno nominal:

```python
build_cpu(num_cores=4, "INTEL", freq=3.1)
```

Python produciría `SyntaxError: positional argument follows keyword argument`.

Es posible especificar **valores por defecto en los parámetros de una función**. En el caso de que no se proporcione un valor al argumento en la llamada a la función, el parámetro correspondiente tomará el valor definido por defecto.

````python
def build_cpu(vendor, num_cores, freq=2.0):
    return {
        "vendor": vendor,
        "num_cores": num_cores,
        "freq": freq,
    }

# Sin especificar frecuencia: se usa el valor por defecto.
print(build_cpu("INTEL", 2))
````

````python
# Indicando una frecuencia concreta.
print(build_cpu("INTEL", 2, 3.4))
````

## Documentación de funciones

Ya hemos visto que en Python podemos incluir comentarios para explicar mejor determinadas zonas de nuestro código.
Del mismo modo podemos (y en muchos casos debemos) adjuntar documentación a la definición de una función incluyendo una cadena de texto (docstring) al comienzo de su cuerpo:

````python
def sqrt(value):
    """Returns the square root of the value."""
    return value ** (1 / 2)

print(sqrt(9))
````

La forma más ortodoxa de escribir un docstring es utilizando triples comillas:

````python
def closest_int(value):
    """Return the closest integer to the given value.

    The operation is:
      1. Compute distance to floor.
      2. If distance is less than a half, return floor.
         Otherwise, return ceil.
    """
    floor = int(value)
    if value - floor < 0.5:
        return floor
    return floor + 1

print(closest_int(3.2))
print(closest_int(3.8))
````

Para ver el docstring de una función, basta con utilizar help:

````python
help(closest_int)
````

También es posible extraer información usando el símbolo de interrogación:

````python
# En Jupyter, el signo ? muestra información de ayuda enriquecida.
# Para que la celda sea Python estándar y reproducible usamos help().
help(closest_int)
````

## Tipos de funciones.

### Funciones anónimas «lambda»

Una función lambda tiene las siguientes propiedades:

1. Se escribe en una única sentencia (línea).
1. No tiene nombre (anónima).
1. Su cuerpo conlleva un *"return"* implícito.
1. Puede recibir cualquier número de parámetros.

Aquí podemos ver un esquema de cómo funcionan.

![funcion_lambda](imagenes/lambda.png)
Veamos un primer ejemplo de función «lambda» que nos permite contar el número de palabras en una cadena de texto dada. La transformación de su versión clásica en su versión anónima sería la siguiente:

````python
num_words = lambda t: len(t.split())
print(type(num_words))
````

````python
print(num_words("hola socio vamos a ver"))
````

### Lambdas como argumentos

Las funciones «lambda» son bastante utilizadas como argumentos a otras funciones. Un ejemplo claro de ello es la función sorted que recibe un parámetro opcional key donde se define la clave de ordenación. Veamos cómo usar una función anónima «lambda» para ordenar una tupla de pares longitud-latitud:

````python
geoloc = (
    (15.623037, 13.258358),
    (55.147488, -2.667338),
    (54.572062, -73.285171),
    (3.152857, 115.327724),
    (-40.454262, 172.318877),
)

# Ordenación por longitud (primer elemento de la tupla).
print(sorted(geoloc))
````

````python
# Ordenación por latitud (segundo elemento de la tupla).
print(sorted(geoloc, key=lambda t: t[1]))
````

## Consejos para programar

1. Las funciones deberían hacer una única cosa.
1. Utiliza nombres descriptivos y con significado.
1. Mantenerlo regularmente y con sentido.
1. No utilices «números mágicos» o valores «hard-codeados».
1. Escribe lo que necesites ahora, no lo que pienses que podrías necesitar en el futuro.
1. Usa comentarios para explicar el «por qué» y no el «qué». (Contextualizar)
